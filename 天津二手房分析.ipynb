```python
# import = 把别人写好的工具箱搬进来用
# pandas 是专门处理表格的库，"as pd" 是给它起个简称，之后写 pd 就行，不用写全拼
import pandas as pd
```

```python
# pd.read_csv() = 读入一个逗号分隔的表格文件，结果存进变量 df
# df 是 dataframe（数据框）的缩写，行内通用叫法，你也可以叫 table，但大家习惯用 df
# 路径前面那个 r 不是笔误：Windows 路径里的反斜杠 \ 会被当成转义符，加 r 表示"照原样读"
# encoding="utf-8-sig" 是编码方式，中文表不加这个，列名第一个字前面会多个看不见的乱码
df = pd.read_csv(
    r"D:\自然语言处理\职业规划\项目01_天津二手房\天津二手房_2026年10月.csv",
    encoding="utf-8-sig"
)
```

```python
# 单独写一个变量名 = 把它的内容显示出来。
# 注意：在 Jupyter 里，一个单元格只显示最后一行；print() 则会原样打出文字
df
```

```python
# df.shape 返回两个数：(行数, 列数)，800 行 11 列说明数据完整读进来了
print("数据有多少行多少列：", df.shape)
print()  # 空 print 就是输出一个空行，只为排版好看

# df.info() 打印每一列的类型和非空数量：
# object = 文字，int64 / float64 = 整数 / 小数
# 某列 Non-Null Count 小于总行数，就说明那列有缺失值
print("每一列都是什么类型、有没有缺失：")
df.info()
```

```python
# df.head(10) 看前 10 行原始数据。括号里不写数字默认看 5 行
# 第一件事永远是"先亲眼看看数据长什么样"，别直接开始算
df.head(10)
```

```python
# df.describe() 一次性给出每列的计数、均值、标准差、最小值、四分位、最大值
# 主要用来快速发现异常：比如面积最小值如果是 0 或 9999，那就是脏数据
df.describe()
```

```python
# groupby("区域") = 按「区域」这一列把 800 行分成若干组（和平区一堆、宝坻区一堆）
# ["单价"]     = 分组之后，只对「单价」这一列做计算
# .mean()      = 每组求平均值；换成 .sum() 求和、.count() 数条数、.max() 取最大
# .sort_values(ascending=False) = 按数值从大到小排（ascending=False 就是降序）
# .round(0)    = 四舍五入保留 0 位小数，只为看着干净
# 这一串是从左往右依次执行的「流水线」：分组 -> 取列 -> 求均值 -> 排序 -> 取整
df.groupby("区域")["单价"].mean().sort_values(ascending=False).round(0)
```

### 问题一：天津各区房价差异有多大

**做法**：按区域分组，算平均单价，从高到低排序。

**结果**：
- 和平区最高，约 60,512 元/㎡
- 宝坻区最低，约 9,645 元/㎡
- 两者相差约 6.3 倍

**结论**
- 房价受行政区域影响较大，和平区平均单价 60512 元/㎡，宝坻区 9645 元/㎡，相差 6.3 倍，呈现明显的“市区—远郊”两套价格体系。因此全市平均单价（约 25252 元）不能代表任何一个具体区域，用它判断买房会严重误判。需要注意的是，这里的差异只反映挂牌均价的差距，并未考虑各区房源数量、面积结构和楼龄，因此不能解释差异的具体成因。
